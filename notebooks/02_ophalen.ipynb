```python
import os
import requests
from datetime import datetime

url = f"https://data.eindhoven.nl/api/explore/v2.1/catalog/datasets/{dataset_id}/exports/csv"
doelmap = "/Volumes/eindhoven/landing/aanlevering/meldingen"
dataset_id = "meldingen-openbare-ruimte-binnenbeter"
bestand = f"{doelmap}/{dataset_id}_{datetime.today().strftime('%Y-%m-%d')}.csv"

os.makedirs(doelmap, exist_ok=True)

if os.path.exists(bestand):
  print(f"Bestand {bestand} bestaat al, overschrijven niet")
else:
  r = requests.get(url, timeout=60)
  r.raise_for_status()
  with open(bestand, "wb") as f:
    f.write(r.content)
  print(f"Bestand {bestand} opgeslagen")
  print("Size (KB):", round(os.path.getsize(bestand)/1024, 2))
  print("Lines:", r.text.count("\n"))
```
