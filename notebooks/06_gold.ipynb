```python
%sql
CREATE OR REPLACE TABLE eindhoven.gold.meldingen_per_wijk
AS SELECT
    wijkcode,
    wijknaam,
    COUNT(*) as aantal_meldingen,
    ROUND(AVG(oplostijd_dagen), 1) AS gem_oplostijd_dagen,
    MEDIAN(oplostijd_dagen) AS mediaan_oplostijd_dagen
FROM
    eindhoven.silver.meldingen
GROUP BY
    wijkcode,
    wijknaam;
    
```

```python
%sql
SELECT * FROM eindhoven.gold.meldingen_per_wijk
ORDER BY aantal_meldingen DESC;
```
